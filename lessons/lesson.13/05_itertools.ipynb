```python
from itertools import zip_longest, combinations, permutations, combinations_with_replacement
```

```python
users = [
    "Bob",
    "John",
    "Alice",
    "Kyle",
    "Leo",
]
```

```python
for num, name in enumerate(users, start=1):
    print(num, name)
```

```python
for num, name in zip(users, range(10, 100, 5)):
    print(num, name)
```

```python
list(zip(users, range(10, 100, 5)))
```

```python
pairs = [('Bob', 10), ('John', 15), ('Alice', 20), ('Kyle', 25), ('Leo', 30)]
list(zip(*pairs))
```

```python
for val, name in zip_longest(range(10, 100, 5), users):
    print(val, name)
```

```python
for val, name in zip_longest(
    range(10, 100, 5),
    users,
    fillvalue="<no-name>",
):
    print(val, name)
```

```python
for pair in combinations(users, 2):
    print(pair)
```

```python
for pair in permutations(users, 2):
    print(pair)
```

```python
for pair in combinations_with_replacement(users, 2):
    print(pair)
```
