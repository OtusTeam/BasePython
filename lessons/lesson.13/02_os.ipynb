```python
for item in os.listdir():
    print("===", item)
    print("isfile:", os.path.isfile(item))
    print("isdir:", os.path.isdir(item))
```

```python
for name in ['pyproject.toml', '.venv', 'foobar.txt']:
    path = os.path.join(os.getcwd(), name)
    print("===", path)
    print("exists:", os.path.exists(path))
    print("isfile:", os.path.isfile(path))
    print("isdir:", os.path.isdir(path))
```

```python
filepath = os.path.join(os.getcwd(), "file.txt")
print("0 exists:", os.path.exists(filepath))
with open(filepath, "w") as f:
    f.write("Hello\n")

print()
print("1 exists:", os.path.exists(filepath))
with open(filepath, "r") as f:
    print("contents:", f.read())

os.unlink(filepath)
print("2 exists:", os.path.exists(filepath))
```
