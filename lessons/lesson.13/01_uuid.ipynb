```python
new_ts_id = uuid.uuid7()
print(new_ts_id)
print(new_ts_id.time)
```

```python
from datetime import datetime, timezone

dt = datetime.fromtimestamp(
    new_ts_id.time / 1000.0,
    tz=timezone.utc,
)
print(dt)
print(dt.timestamp())
```

```python
uuids = [uuid.uuid7() for _ in range(1000)]
```

```python
print(uuids[0])
print(uuids[-1])
```

```python
def extract_uuid7_counter(u: uuid.UUID) -> int:
    # Сдвигаем 128-битное число вправо, чтобы отбросить последние биты (вариант и рандом)
    # И маскируем первые биты (timestamp и версию), оставляя только 42 бита счетчика
    return (u.int >> 34) & 0x3FFFFFFFFFF
```

```python
print(extract_uuid7_counter(uuids[0]))
print(extract_uuid7_counter(uuids[-1]))
```
