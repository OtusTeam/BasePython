```python
filepath = Path.cwd() / "file.txt"
filepath
```

```python
print("exists:", filepath.exists())
print("is_file:", filepath.is_file())
```

```python
filepath.write_text("hello world!\n")
print("exists:", filepath.exists())
print("is_file:", filepath.is_file())
```

```python
print("contents:", filepath.read_text())
```

```python
filepath.unlink(missing_ok=True)
print("exists:", filepath.exists())
print("is_file:", filepath.is_file())
```

```python
filepath.suffix
```

```python
cats_pics = Path.cwd() / "pics" / "cats"
print(cats_pics.exists())
print(cats_pics.is_dir())
cats_pics.mkdir(parents=True, exist_ok=True)
print(cats_pics.exists())
print(cats_pics.is_dir())
```
