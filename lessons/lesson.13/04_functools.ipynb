```python
from functools import wraps, cache, partial
```

```python
from timeit import default_timer as timer


def show_timing(func):
    @wraps(func)
    def wrapper(*a, **kwargs):
        """
        hello timing wrapper
        :param a:
        :param kwargs:
        :return:
        """
        start_time = timer()
        try:
            return func(*a, **kwargs)
        finally:
            total_time = timer() - start_time
            print("total time:", total_time)

    return wrapper
```

```python
@show_timing
def mul(a, b):
    """
    multiplier
    :param a:
    :param b:
    :return:
    """
    return a * b

print(mul)
print(mul.__wrapped__)
```

```python
def factorial(n):
    print(f"get fac({n})")
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

```python
factorial(1)
```

```python
factorial(10)
```

```python
factorial(100)
```

```python
@cache
def factorial(n):
    print(f"get fac({n})")
    if n == 0:
        return 1
    return n * factorial(n - 1)
```

```python
@show_timing
@cache
def factorial(n):
    print(f"get fac({n})")
    if n == 0:
        return 1
    return n * factorial(n - 1)
```
