```python
from collections import defaultdict
import uuid

# 1. Генерируем 1000 UUIDv7
uuids = [uuid.uuid7() for _ in range(1000)]

timeline = defaultdict(list)

for u in uuids:
    # Приводим к строке: xxxxxxxx-xxxx-7xxx-[yyyy]-[zzzzzzzzzzzz]
    # Нам нужны сегменты: timestamp (первые два) и блоки случайных данных/счетчика
    parts = str(u).split('-')

    # 48 бит timestamp собираются из первых двух частей
    timestamp_ms = int(parts[0] + parts[1], 16)

    # Извлекаем сегменты счетчика и случайных данных:
    # parts[2] — это '7xxx' (4 бита версии '7' + 12 бит счетчика/рандома)
    # parts[3] — это 'yyyy' (2 бита варианта + 14 бит счетчика/рандома)
    # parts[4] — это 'zzzzzzzzzzzz' (оставшийся рандом)

    # Для точной визуализации склеим ВСЕ биты после версии и варианта в одно число
    # Отбрасываем '7' из parts[2] и берем чистые данные
    clean_rand_a = parts[2][1:] # 12 бит (3 hex символа)
    clean_rand_b = parts[3]     # 16 бит (4 hex символа, включая вариант)
    clean_rand_c = parts[4]     # 48 бит (12 hex символов)

    # Переводим в число всю последовательность, которая должна инкрементироваться
    full_counter = int(clean_rand_a + clean_rand_b + clean_rand_c, 16)

    timeline[timestamp_ms].append(full_counter)

# 3. Вывод сгруппированного результата
print(f"Всего генерация заняла: {len(timeline)} мс\n")

for ms, counters in timeline.items():
    print(f"⏱️ Миллисекунда (Unix TS): {ms}")
    print(f"   └─ Сгенерировано UUID: {len(counters)} шт.")
    print(f"   └─ Первые 5 счетчиков (Full): {counters[:5]}")

    is_unique_inside_ms = len(counters) == len(set(counters))
    if is_unique_inside_ms:
        print("   └─ Status: ✅ Все счетчики внутри этой миллисекунды строго уникальны!")
    else:
        print("   └─ Status: ❌ Все еще есть дубликаты.")
    print("-" * 50)
```
