```python
password = int(input("Введите пароль: "))

if password == 1234:
    print("Доступ получен")
else:
    password = int(input("Введите пароль: "))
    if password == 1234:
        print("Доступ получен")
```

```python
password = int(input("Введите пароль: "))

while not password == 1234:
    print("Доступ запрещен. Введите еще раз пароль")
    password = int(input("Введите пароль: "))
print("Доступ получен")
```
